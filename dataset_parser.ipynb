```python
import pandas as pd
```

```python
#Variables con las rutas de los archivos para que solamente
#Sea cambiar el indice de la ruta y que el programa haga el resto
base_route = "./csv/excel/"
xlsx_paths = [{"out_path": "calc_case_train.xlsx", "path": "calc_case_description_train_set.xlsx"},
             {"out_path": "calc_case_test.xlsx", "path": "calc_case_description_test_set.xlsx"},
             {"out_path": "mass_case_train.xlsx", "path": "mass_case_description_train_set.xlsx"},
             {"out_path": "mass_case_test.xlsx", "path": "mass_case_description_test_set.xlsx"}]

selected_item = xlsx_paths[3]
df = pd.read_excel(base_route+selected_item["path"])
df.head()
```

```python
dicom_path = base_route + "dicom_info.xlsx"
dicom_df = pd.read_excel(dicom_path)
dicom_df.head()
```

```python
items = []
for _, row in df.iterrows():
    full_mammogram_id = row["image file path"].split("/")[0]
    cropped_image_id = row["cropped image file path"].split("/")[0]
    roi_image_id = row["ROI mask file path"].split("/")[0]
    target = row["pathology"]
    
    fila_fm = dicom_df[
    (dicom_df["PatientID"] == full_mammogram_id) &
    (dicom_df["SeriesDescription"] == "full mammogram images")].iloc[0]

    fila_ci = dicom_df[
    (dicom_df["PatientID"] == cropped_image_id) &
    (dicom_df["SeriesDescription"] == "cropped images")].iloc[0]

    fila_roi = dicom_df[
    (dicom_df["PatientID"] == roi_image_id) &
    (dicom_df["SeriesDescription"] == "ROI mask images")].iloc[0]

    items.append([full_mammogram_id, fila_fm["image_path"], cropped_image_id, fila_ci["image_path"], roi_image_id, fila_roi["image_path"], row["patient_id"], target])
columns = ["full_mammogram_id", "full_mammogram_img_path", "cropped_image_id", "cropped_image_img_path", "roi_image_id", "roi_image_img_path","patient_id", "pathology"]
parsed_df = pd.DataFrame(items, columns=columns)

parsed_df.head()
```
